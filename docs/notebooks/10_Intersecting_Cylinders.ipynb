# Intersecting Cylinders

Curved-fold tessellations from tilings whose face incenters are mutually tangential.
Each face becomes a curved 'bump' that meets its neighbours along the edges of the
original tiling; with the canonical quarter-circle profile on the platonic 4 (square)
tiling the result is a tessellation of intersecting half-cylinders.

The crease pattern inserts a small twist at every original vertex and uses straight
and curved creases between the twists and the face incenters. With ``r=1`` curved
triangles meet at the incenters; with ``0 < r < 1`` a downscaled copy of the
original face is preserved and the triangles become curved quadrilaterals.

See :mod:`eucare.intersecting_cylinders` for the API.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

import eucare as ec
from eucare.intersecting_cylinders import (
    Profile,
    circular_profile,
    spherical_profile,
    make_intersecting_cylinders,
)
```

## Cross-section profiles

A `Profile` describes the shape of the curved triangles perpendicular to their flat
edge. `circular_profile()` is the canonical quarter-ellipse; `spherical_profile()`
stays flat near the edge and rises like a sphere's pole.

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
circular_profile(scale=1.3).plot(ax=axes[0])
axes[0].set_title('circular_profile(scale=1.3)')
spherical_profile(flat_until=0.6).plot(ax=axes[1])
axes[1].set_title('spherical_profile(flat_until=0.6)')
plt.tight_layout()
plt.show()
```

## Building the crease pattern

Provide any tiling whose faces have well-defined (pseudo-)incenters. Below we use
the platonic 4 tiling (squares).

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=2)
CP = make_intersecting_cylinders(G, circular_profile(), r=1.0)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```

## Preserving the original faces with `r < 1`

With `r < 1`, a downscaled copy of every original face is kept and the curved
triangles become curved quadrilaterals reaching down to it.

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=2)
CP = make_intersecting_cylinders(G, circular_profile(), r=0.7)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```

## Spherical profile on a triangular tiling

The spherical profile produces dome-like bumps with a flat skirt around each face.

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(3), rings=2)
CP = make_intersecting_cylinders(G, spherical_profile(flat_until=0.5), r=1.0)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```

## Custom profiles

Any function `f: [0, 1] -> R` with `f(0) = 0` works via `Profile.from_function`.
Below we use a parabolic bump.

```python
parabolic = Profile.from_function(lambda x: 1 - (1 - x) ** 2)
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(6), rings=2)
CP = make_intersecting_cylinders(G, parabolic, r=1.0)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```
