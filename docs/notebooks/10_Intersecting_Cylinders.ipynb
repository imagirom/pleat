# Intersecting Cylinders

Curved-fold tessellations from tilings whose face incenters are mutually tangential.
Each face becomes a curved 'bump' that meets its neighbours along the edges of the
original tiling; with the canonical quarter-circle profile on the platonic 4 (square)
tiling the result is a tessellation of intersecting half-cylinders.

The crease pattern inserts a small twist at every original vertex and uses straight
and curved creases between the twists and the face incenters. With ``r=1`` curved
triangles meet at the incenters; with ``0 < r < 1`` a downscaled copy of the
original face is preserved and the triangles become curved quadrilaterals.

See :mod:`eucare.intersecting_cylinders` for the API.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

import eucare as ec
from eucare.intersecting_cylinders import (
    Profile,
    circular_profile,
    make_intersecting_cylinders,
    show_3d,
    top_view,
)
```

## Cross-section profiles

A `Profile` describes the shape of the curved triangles perpendicular to their flat
edge. The default `circular_profile(scale=1.3)` is a tall quarter-ellipse;
`circular_profile(scale=1.0)` is the canonical quarter-circle that yields exact
intersecting half-cylinders on the platonic 4 and 3 tilings.

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
circular_profile(scale=1.0).plot(ax=axes[0])
axes[0].set_title('circular_profile(scale=1.0)')
circular_profile(scale=1.3).plot(ax=axes[1])
axes[1].set_title('circular_profile(scale=1.3)')
plt.tight_layout()
plt.show()
```

## Building the crease pattern

Provide any tiling whose faces have well-defined (pseudo-)incenters. Below we
use the platonic 4 tiling (squares) with a tall quarter-ellipse profile.

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=2)
CP = make_intersecting_cylinders(G, circular_profile(), r=1.0)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```

### Top-view projection

The flat top-view of the folded model is obtained by joining each edge to the
corresponding face incenter (the Conway `join` operator):

```python
tv = top_view(G, r=1.0)
tv.show(line_width=0.02, render_vertices=False, render_faces=True)
```

## Preserving the original faces with `r < 1`

With `r < 1`, a downscaled copy of every original face is kept and the curved
triangles become curved quadrilaterals reaching down to it. The corresponding
top view is the Conway `chamfer` operator with parameter `t = r`.

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=2)
CP = make_intersecting_cylinders(G, circular_profile(), r=0.7)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=2)
top_view(G, r=0.7).show(line_width=0.02, render_vertices=False, render_faces=True)
```

## Interactive 3D preview

`show_3d` builds a triangle mesh of the folded surface and returns a
[plotly](https://plotly.com/python/) figure that can be rotated, zoomed and
panned directly in the notebook. With the canonical `circular_profile(scale=1.0)`
on the platonic 4 tiling we recover the eponymous *intersecting half-cylinders*.

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(4), rings=1)
show_3d(G, circular_profile(scale=1.0), r=1.0)
```

### 3D preview with `r < 1`

For `r < 1` the curved patches end on the lifted, shrunken inner faces:

```python
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(3), rings=2)
show_3d(G, circular_profile(scale=1.3), r=0.6)
```

## Custom profiles

Any function `f: [0, 1] -> R` with `f(0) = 0` works via `Profile.from_function`.
Below we use a parabolic bump.

```python
parabolic = Profile.from_function(lambda x: 1 - (1 - x) ** 2)
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(6), rings=2)
CP = make_intersecting_cylinders(G, parabolic, r=1.0)
CP.show(line_width=0.02, render_vertices=False, render_faces=False)
```
