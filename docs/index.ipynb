# Pleat

Welcome to **pleat**, a Python library for constructing [geometric tilings](https://en.wikipedia.org/wiki/Tessellation) and turning them into [crease patterns](https://en.wikipedia.org/wiki/Crease_pattern) which can be folded into [origami tessellations](https://origamiusa.org/thefold/article/tessellations-brief-theory-warping-paper). It supports Euclidean, hyperbolic, and spherical geometries, provides a suite of Conway topological operators, and can generate crease patterns with computed folded states.

The heart of this documentation are the notebooks, which guide you through many of the features — this home page is the first of them. It walks through the *whole* origami pipeline in a few cells:

1. **Build a tiling** — Construct a tiling using one of the many provided methods ([Euclidean](notebooks/01_Tilings_Euclidean.ipynb), [spherical & hyperbolic](notebooks/02_Curved_Geometries.ipynb), [Conway operators](notebooks/05_Conway_Operators.ipynb)), or import your own — from an SVG or even a [hand-drawn image](notebooks/Image_To_Graph.ipynb).
2. **Construct a crease pattern** — apply an algorithm such as [*shrink-rotate*](notebooks/Shrink_Rotate_Tessellations.ipynb) to construct a crease pattern (CP) from the tiling geometry.
3. **Fold and export** — Generate a preview of the folded state and [export the results](notebooks/Saving_and_Exporting.ipynb), for viewing, printing or plotting.

```python
import pleat as ec
```

## Step 1: build a tiling

```python
# create a tiling to work with: Two rings of hexagons around a central hexagon
# Try the other platonic tilesets with n=3 and n=4 as well!
G = ec.example_graphs.from_tiles(ec.example_tilesets.platonic(n=6), rings=2)
G.show()
```

## Step 2: construct a crease pattern

```python
from pleat.flat_foldable import max_kawasaki_sum
from pleat.shrink_rotate import assign_this_way_by_bfs, shrink_rotate_pattern
import numpy as np

# makes it so that in the folded form, the central face will lie on top of all its neighbors, which lie on top of their further neighbors, etc.
assign_this_way_by_bfs(G, G.central_face())  

# construct the crease pattern with the shrink-rotate method
CP = shrink_rotate_pattern(G, simplify_boundary=True, alpha=np.pi/5, factor=0.5)  
CP.show(**ec.rendering.CREASE_PATTERN_PRESET)

# confirm that Kawasaki's theorem is satisfied at each vertex, which is a necessary condition for flat foldability
assert max_kawasaki_sum(CP) < 1e-10 
```

## Step 3: fold and export

```python
# fold the crease pattern, generating a backlit view and solving the layer ordering
fold_result = ec.overlap.fold_complete(CP, quiet=True)

# display the folded form along with the CP
fold_result.show()

# uncomment and specify a path to export the crease pattern and the folded form to a folder
# output_folder = 
# fold_result.save(output_folder)
```
